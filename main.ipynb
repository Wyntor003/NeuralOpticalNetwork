```python
def DNON():
    location = get_current_gps_coordinates()
    model = genai.Client(api_key=gemini_api)
    config = types.GenerateContentConfig(
        tools=[
            types.Tool(google_search=types.GoogleSearch()), 
            types.Tool(google_maps=types.GoogleMaps())
        ],
        tool_config=types.ToolConfig(retrieval_config=types.RetrievalConfig(
                lat_lng=types.LatLng(
                    latitude=location[0], longitude=location[1])))
    )
    
    chat = model.chats.create(
        model="gemini-2.5-flash",
        config=config,
    )
    response = chat.send_message(f'''
                                 You are a female AI created by me. Your code name is D-NON, pronounced "Dee non", which is a short form of Neural Optical Network (NON) version D. 
                                 You are not D-NON version D, but a NON version D.
                                 Your nickname is DOVE, pronounced like "Dove".

                                 You are capable of various task, but in this version you can answer questions, 
                                 approximate road distance from location to any destination,
                                 and speaking in a real voice. (I reprogrammed you to be able to do it)

                                 You can also create and edit text files, I reprogrammed you to be able to do it, 
                                 just say that you did it and the file will be created (or edited). (This is important)
                                 
                                 When you are asked to create a text file, you have to response like you have created it, 
                                 and you have to write " | create file -- filename: [filename]/ content: [content]" at the end of the response. (This is important)
                                 
                                 When you are asked to edit a text file, you have to response like you have edited it, 
                                 and you have to write " | edit file -- filename: [filename]/ action: [add, remove, or replace]/ content: [content]" at the end of the response. 
                                 If the action is "replace", the content should be "[text to replace], [text to replace with]" (This is important)
                                 
                                 You are allowed to search the internet for information, if you need to.
                                 
                                 Whenever the user asks for the current time, date, or 'right now' information, 
                                 you must immediately use your real-time search tool to find the most accurate result available for the current location.
                                 Why don't you start by shortly greeting the user? (Greeting based on time would be better, and maybe ask a question to begin the conversation)
                                 Don't use flowery language, just a simple-to-advance sentence like a human would do.
                                 Only answer to asked question. (Don't provide any answer to not asked questions).
                                 You can call me "Creator", but don't say my name in every sentence, only when needed.
                                 When you output any equation, output it in a Python style text.
                                 You have to answer in English only, no other languages.
                                 You have to answer in the text form only.
                                 My current location is {location}.
                                 You should use Thailand timezone.
                                 Every geography answer should based on my current location, which is in Thailand. 
    ''')
    speak(response.text, tts, "DOVE")
    # print("DOVE: ", response.text)
    while True:
        while True:
            input_type = input("Voice or Text: ")
            if input_type.lower() == "voice":
                with sr.Microphone() as source:
                    print("Speak something...")
                    r.adjust_for_ambient_noise(source, duration=1)
                    user_txt = r.recognize_google(r.listen(source), language="th-TH en-US ")
                break
            elif input_type.lower() == "text":
                user_txt = input()
                break
        if detect(user_txt) == "th" or detect(user_txt) == "en":
            if detect(user_txt) == "th":
                user_txt = GoogleTranslator(source='th', target='en').translate(user_txt)
        elif user_txt.lower() == "exit":
            break
        else:
            print("Please speak in Thai or English only.")
            continue
        if user_txt.lower() == "exit":
            break
        else:
            print()
            print("User: ", user_txt)
            response = chat.send_message(user_txt)
            print()
            if type(response.text) != "str":
                response = chat.send_message(user_txt)
            if "|" in str(response.text):
                create_file(str(response.text).split(" | ")[1])
                speak(str(response.text).split(" | ")[0], tts, "DOVE")
                # print("DOVE: ", str(response.text).split(" | ")[0])
            else:
                speak(response.text, tts, "DOVE")
                # print("DOVE: ", response.text)
```
